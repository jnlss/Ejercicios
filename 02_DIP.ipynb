La DIP (Inyección de Dependencias) es uno de los principios SOLID (la D, para ser más precisos). La máxima de este principio es que __los módulos de alto nivel no deben depender de los modulos de bajo nivel. Ambos deben depender de abstracciones__.

Veamos un ejemplo:

```python
# Simulación de la clase Motor
class Motor:
    def __init__(self):
        self.presion_pedal = 0

    def torque(self):
        # Simulación simple del cálculo de torque
        return self.presion_pedal * 10


class Vehiculo:
    def __init__(self):
        self.motor = Motor()  # Creación interna del motor
```

En este caso, la clase Vehículo (alto nivel) depende directamente de la clase Motor (bajo nivel), ya que se utiliza su constructor dentro.

Como alternativa, el principio de inyección de dependencias dice que ambas clases deben depender únicamente de abstracciones. 

Para empezar, vamos a modificar la clase Vehículo para que no dependa del constructor:

```python
class Vehiculo:
    def __init__(self, motor:MotorBase =None):
        self.motor = motor
```

Ahora, un cambio en el constructor de Motor no rompe el código de vehículo. Podemos aumentar el desacoplamiento creando una interfaz, de manera que la clase Vehículo se comunique con la clase Motor solo a través de ella.

```python

# Abstracción del motor
class MotorBase(ABC):
    @abstractmethod
    def torque(self):
        pass

# Implementación concreta del motor
class MotorCombustion(MotorBase):
    def __init__(self):
        self.presion_pedal = 0

    def torque(self):
        # Simulación simple del cálculo de torque
        return self.presion_pedal * 10
```
