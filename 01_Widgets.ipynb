# Introducción a tkinter

Esta biblioteca permite crear interfaces gráficas sencillas. Como veréis, las interfaces van a tener un aspecto muy... noventero. Existen herramientas más modernas y sofisticadas para desarrollar aplicaciones que parezcan de este siglo, pero esta librería tiene dos ventajas: es la predeterminada de Python y es muy sencilla de utilizar.

Comencemos importando los paquetes: tkinter es el principal, y dentro de él está el módulo ttk. Este módulo comparte muchas funcionalidades con el módulo raíz de tkinter, pero ofrece soluciones más modernas.

```python
import tkinter as tk
from tkinter import ttk
```

Llegó la hora de hacer un Hola Mundo con tkinter. Para ello, necesitamos dos componentes: la clase raíz que encapsula todas las funcionalidades (el objeto root) y el campo de texto.

```python
# Creamos el objeto raíz. En él se incluirán todos los objetos que queramos introducir en nuestra app
root = tk.Tk()
# Presentamos el objeto label: un campo de texto. Su primer argumento (y el de cualquier widget) es el directorio raíz
label = ttk.Label(root, text="Hello World!")
# El método pack() hace que se muestre el widget en el contenedor root
# Tiene argumentos que permiten moderar cómo se muestra
label.pack()
# Esta última línea la debemos poner siempre. Hace que el objeto root se ejecute en bucle
root.mainloop()
```

# Encapsular los elementos en frames

Normalmente, las interfaces gráficas se encapsulan en zonas para no tener todo dentro de un mismo cuadro. Esto hace que los componentes sean más controlables y da una noción de jerarquía a la aplicación.

Este encapsulamiento se hace mediante la clase Frame. Vamos a ver el ejemplo anterior de manera más organizada:

```python
root = tk.Tk()
frame = ttk.Frame(root)
frame.pack()
# Ahora, el label está dentro del frame y no de root
ttk.Label(frame, text="Hello World!").pack()
ttk.Label(frame, text="Bye World!").pack()
root.mainloop()
```

Ya puestos, vamos a explicar una alternativa al pack: el grid. Ambos tienen la misma función: habilitar el Widget dentro del espacio (bien sea un Frame, root...). Permite una organización más limpia en cuadrículas.

```python
root = tk.Tk()
frame = ttk.Frame(root)
frame.pack()

# Con grid, podemos invertir fácilmente el orden de los elementos en vez de depender del orden aparición
ttk.Label(frame, text="Hello World!", font=100).grid(column=0, row=1)
ttk.Label(frame, text="Bye World!", font=100).grid(column=0, row=0)
root.mainloop()
```

Aparte de pack y grid, también existe place, pero es para colocar los elementos en el espacio de coordenadas.

# Encapsulamiento en clases

Las aplicaciones normalmente son muy grandes y cuentan con un amplio abanico de frames que se vuelven visibles y se cierran. Imagina introducir todos los componentes de una aplicación como hemos hecho hasta ahora... ¡sería una pesadilla!

Por suerte, una vieja amiga ha venido a salvarnos: la Programación Orientada a Objetos va a permitirnos jerarquizar, encapsular y modularizar el código. Vamos a crear una clase que herede de Frame para introducir el mismo contenido de antes de manera más ordenada:

```python
class App(tk.Frame):
    # master es el root de antes, el componente padre en el que va a ir encapsulado este Frame
    def __init__(self, master):
        super().__init__(master)
        # Realizamos la misma lógica que antes
        self.pack()
        
        # El primer argumento de Label es el Frame en el que se encapsula dicho Label. En este caso, es self.
        self.helloLabel = ttk.Label(self, text="Hello World!",font=50)
        self.byeLabel = ttk.Label(self, text="Bye World!", font=50)
        
        self.helloLabel.grid(row=0)
        self.byeLabel.grid(row=1)
```

Bastante más organizado, ¿verdad? Ahora tenemos los dos Labels inicializados dentro de su ventana en vez de en nuestro programa principal. Vamos a iniciar nuestra aplicación como antes:

```python
root = tk.Tk()
app = App(root)
root.mainloop()
```

# Widgets disponibles

En las diapositivas tenéis un amplio abanico de los widgets disponibles en tkinter. Algunos se usan más esporádicamente que otros. Para finalizar este notebook, vamos a crear una pequeña aplicación Hola Mundo que utilice nuestro nombre. Necesitaremos tres elementos:
- Un Label en el que se despliegue nuestro nombre con un hola mundo.
- Una entrada de texto (Entry) donde introduzcamos nuestro nombre.
- Un botón que actualice el Label cuando le pulsemos.

Empecemos por la entrada de texto y el Label, que es lo más sencillo:

```python
class App(tk.Frame):
    def __init__(self, master):
        super().__init__(master)
        # Realizamos la misma lógica que antes
        self.pack()
        
        # El primer argumento de Label es el Frame en el que se encapsula dicho Label. En este caso, es self.
        self.nameLabel = ttk.Label(self, text="Inserta tu nombre:", font=50)
        self.nameEntry = ttk.Entry(self, font=50)
        
        # Con el método get obtenemos el valor introducido en nameEntry
        self.helloLabel = ttk.Label(self, text="Hola," + self.nameEntry.get(), font=50)
        
        # Posicionamos los widgets
        self.nameLabel.grid(row=0, column=0)
        self.nameEntry.grid(row=0, column=1)
        self.helloLabel.grid(row=1)
```

```python
root = tk.Tk()
app = App(root)
root.mainloop()
```

Como vemos, con el método get() podemos extraer el texto que haya en nameEntry... pero solo en el momento de creación. Necesitamos el botón para refrescar el texto. Para ello, creamos un nuevo método dentro de la función, y el botón llamará a ese método cuando sea pulsado.

```python
class App(tk.Frame):
    def __init__(self, master):
        super().__init__(master)
        self.pack()
        
        self.nameLabel = ttk.Label(self, text="Inserta tu nombre:", font=50)
        self.nameEntry = ttk.Entry(self, font=50)
        
        self.helloLabel = ttk.Label(self, text="Hola," + self.nameEntry.get(), font=50)
        
        # Creación del botón: la opción command nos permite asociar una función al click
        self.button = ttk.Button(self, text="Refrescar", command=self.refrescaNombre)
        
        # Posicionamos los widgets
        self.nameLabel.grid(row=0, column=0)
        self.nameEntry.grid(row=0, column=1)
        self.helloLabel.grid(row=1)
        self.button.grid(row=2)
        
    def refrescaNombre(self):
        self.helloLabel.config(text="Hola," + self.nameEntry.get())
```

```python
root = tk.Tk()
app = App(root)
root.mainloop()
```

Si te has fijado, en command hemos puesto refresaNombre sin paréntesis. Eso quiere decir que no estamos pasando como argumento el resultado de la función... ¡sino la propia función! Por dentro, la clase Button ya ejecutará la función cuando le convenga (que será cuando pulsemos el botón).

Ahora cobra sentido una frase que os he repetido unas cuantas veces: en Python, todo son objetos, incluso las funciones. Esto hace que también podamos pasar funciones como argumentos de funciones.

Para acabar, vamos a mejorar un poco la estética de la aplicación:
- El texto solo se mostrará tras pulsar por primera vez el botón, para evitar que aparezca vacío.
- Vamos a centrar el texto.
- Vamos a añadir un botón que cierre el Frame.

```python
class App(tk.Frame):
    def __init__(self, master):
        super().__init__(master)
        self.pack()
        
        self.nameLabel = ttk.Label(self, text="Inserta tu nombre:", font=50)
        self.nameEntry = ttk.Entry(self, font=50)
        
        # Inicialmente, el texto es vacío. Si lo queremos hacer desaparecer, existen las funciones:
        # pack_forget, grid_forget
        # anchor='center' nos centra el texto
        self.helloLabel = ttk.Label(self, text="", anchor='center', font=50)
        
        self.button = ttk.Button(self, text="Refrescar", command=self.refrescaNombre)
        # La función destroy cierra el Frame
        # Como lo que queremos es que se cierre la aplicación, tenemos que llamar al destroy de root
        self.quitButton = ttk.Button(self, text="Cerrar", command=self.master.destroy)
        
        # Posicionamos los widgets
        self.nameLabel.grid(row=0, column=0)
        self.nameEntry.grid(row=0, column=1)
        # Con columnspan=2, le decimos a tkinter que este widget ocupe las dos columnas existentes
        self.helloLabel.grid(row=1, columnspan=2)
        
        self.button.grid(row=2, column=0)
        self.quitButton.grid(row=2, column=1)
        
    def refrescaNombre(self):
        self.helloLabel.config(text="Hola," + self.nameEntry.get())
```

```python
root = tk.Tk()
app = App(root)
root.mainloop()
```
